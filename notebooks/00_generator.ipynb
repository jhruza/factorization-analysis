# Inspect the canonical generator

A small deterministic sample from the package generator. This notebook is a sanity check; training data are sampled by the training code.

```python
from factorization.processes import canonical_process

process = canonical_process()
batch = process.sample(batch_size=4, seq_len=8, seed=7, with_truth=True)
print('factor sizes:', process.sizes)
print('ordinary vocabulary:', process.vocab_size, '| BOS:', process.bos_token)
print('tokens:', batch.tokens.shape)
print('targets:', batch.targets.shape)
print('number of factor belief arrays:', len(batch.beliefs))
print('number of factor predictive arrays:', len(batch.factor_probs))
```

```python
# Check that the public evaluator returns the corresponding truth quantities.
truth = process.evaluate(batch.tokens)
print('belief factors:', len(truth.beliefs))
print('predictive factors:', len(truth.factor_probs))
```
